```python
import os
import numpy as np
import pickle
```

# Pool the rPPG fusion waveforms

```python
waveform_files = [
    "fold_0_rgb_left_nir_fusion/pred.pickle",
    "fold_1_rgb_left_nir_fusion/pred.pickle",
    "fold_2_rgb_left_nir_fusion/pred.pickle",
    "fold_3_rgb_left_nir_fusion/pred.pickle"
    ]
all_pred = []
all_gt = []
all_participant_task_chunk_list = []
for waveform_file in waveform_files:
    with open(waveform_file, 'rb') as f:
        data = pickle.load(f)
    all_pred.extend(data['pred'])
    all_gt.extend(data['gt'])
    all_participant_task_chunk_list.extend(data['participant_task_chunk_id_list'])
all_pred = np.array(all_pred)
all_gt = np.array(all_gt)
all_participant_task_chunk_list = np.array(all_participant_task_chunk_list)
print(all_pred.shape, all_gt.shape, all_participant_task_chunk_list.shape)
print(all_pred.dtype, all_gt.dtype, all_participant_task_chunk_list.dtype)
```

Concatenate the waveforms from the same trial/sample

```python
pred_waveform = {participant_task: [] for participant_task, chunk in all_participant_task_chunk_list}
gt_waveform = {participant_task: [] for participant_task, chunk in all_participant_task_chunk_list}
for pred, gt, participant_task_chunk_id in zip(all_pred, all_gt, all_participant_task_chunk_list):
    participant_task = participant_task_chunk_id[0]
    chunk_id = participant_task_chunk_id[1]
    pred_waveform[participant_task].extend(pred.tolist())
    gt_waveform[participant_task].extend(gt.tolist())
for participant_task in pred_waveform:
    pred_waveform[participant_task] = np.array(pred_waveform[participant_task])
    gt_waveform[participant_task] = np.array(gt_waveform[participant_task])
rppg_waveforms = {}
for participant_task in pred_waveform:
    rppg_waveforms[participant_task] = {
        'pred': pred_waveform[participant_task],
        'gt': gt_waveform[participant_task]
    }
print(rppg_waveforms.keys())
print(len(rppg_waveforms.keys()))
```

Save

```python
output_file = "rppg_waveforms.pickle"
with open(output_file, 'wb') as f:
    pickle.dump(rppg_waveforms, f)
```
